```python
## feature engineering process
final_df['Date']=final_df['Date_of_Journey'].str.split('/').str[0]
final_df['Month']=final_df['Date_of_Journey'].str.split('/').str[1]
final_df['Year']=final_df['Date_of_Journey'].str.split('/').str[2]
final_df.head()
```

```python
## convert hour to minute
final_df['duration_hour']=final_df['Duration'].str.split(' ').str[0].str.split('h').str[0]
final_df.head()
```

```python
final_df[final_df['duration_hour']=='5m']
```

```python
def extract_duration_parts(duration_str):
    hours = 0
    minutes = 0
    duration_str = str(duration_str) # Ensure it's a string to prevent errors with non-string types

    # Check for hours
    if 'h' in duration_str:
        h_part = duration_str.split('h')[0].strip()
        try:
            hours = int(h_part)
        except ValueError:
            pass # If conversion fails, hours remains 0

        # Check for minutes after 'h'
        if 'm' in duration_str.split('h')[-1]:
            m_part = duration_str.split('h')[-1].split('m')[0].strip()
            try:
                minutes = int(m_part)
            except ValueError:
                pass # If conversion fails, minutes remains 0
    # Check for minutes if no 'h' was found in the main part
    elif 'm' in duration_str:
        m_part = duration_str.split('m')[0].strip()
        try:
            minutes = int(m_part)
        except ValueError:
            pass # If conversion fails, minutes remains 0

    return hours, minutes

# Apply the function to create new 'duration_hour' and 'duration_min' columns
final_df[['duration_hour', 'duration_min']] = final_df['Duration'].apply(lambda x: pd.Series(extract_duration_parts(x)))

# Ensure they are of integer type (though the function already tries to convert them)
final_df['duration_hour'] = final_df['duration_hour'].astype(int)
final_df['duration_min'] = final_df['duration_min'].astype(int)

# Drop the original 'Duration' column
final_df.drop('Duration', axis=1, inplace=True)
```

```python
final_df.head()
```

```python
# Reconstruct the 'Total_Stops' column from df_train and df_test
# First, concatenate the 'Total_Stops' columns from the original dataframes
total_stops_series = pd.concat([df_train['Total_Stops'], df_test['Total_Stops']], ignore_index=True)

# Define the mapping dictionary
stop_mapping = {'non-stop': 0, '1 stop': 1, '2 stops': 2, '3 stops': 3, '4 stops': 4}

# Apply the mapping to the reconstructed series
final_df['Total_Stops'] = total_stops_series.map(stop_mapping)

# Fill NaN values. The previous agent's attempt mapped 'nan' to 1,
# so we assume this is the desired behavior for any unmapped or original NaN values.
final_df['Total_Stops'].fillna(1, inplace=True)

# Convert to integer type
final_df['Total_Stops'] = final_df['Total_Stops'].astype(int)

# Display head and info to verify
display(final_df.head())
final_df.info()
```
