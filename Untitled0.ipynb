step 1

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score, roc_curve

#xgboost
import xgboost as xgb
from xgboost import XGBClassifier

print(f"XGBoost Version: {xgb.__version__}")
```

step 2

```python
#load dataset
housing = fetch_california_housing(as_frame=True)
df = housing.frame
df.head()
```

```python
# Load dataset to ensure 'MedHouseVal' is present
housing = fetch_california_housing(as_frame=True)
df = housing.frame

#convert into binary classification problem
median_val = df['MedHouseVal'].median()
df['AboveMedian'] = (df['MedHouseVal'] > median_val).astype(int)
df = df.drop(columns=['MedHouseVal'])

#feature engineering
df['AvgOccupantsPerRoom'] = df['AveOccup'] / df['AveRooms']
df['Income_vs_Age'] = df['MedInc'] * df['HouseAge']

la_coords = (34.0522, -118.2437)
df['Distance_to_LA'] = np.sqrt((df['Latitude'] - la_coords[0])**2 + (df['Longitude'] - la_coords[1])**2)

df.head()
```

step 3

```python
#train , test split with stratification
X = df.drop(columns=['AboveMedian'])
y = df['AboveMedian']

# Stratify split ensures equal distribution of the target classes in train/test splits
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
print(f"Train Shape: {X_train.shape} | Test Shape: {X_test.shape}")
```

step 4

```python

#Define and Train XGBoost with Early Stopping
model = XGBClassifier(
    n_estimators=1000,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric='logloss',
    random_state=42,
    tree_method='hist',
    early_stopping_rounds=30  #Defined here instead of inside .fit()
)

# Fit model with evaluation set
model.fit(
    X_train, y_train,
    eval_set=[(X_train, y_train), (X_test, y_test)],
    verbose=50
)
```

step 5

```python
#predictions and evalution
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

# Print evaluation metrics
print("--- Classification Report ---")
print(classification_report(y_test, y_pred))
print(f"ROC-AUC Score: {roc_auc_score(y_test,y_prob):.4f}")
```

step 6

```python
#visualization and feature metrics
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
sns.set_theme(style="whitegrid")

#training vs validation
results = model.evals_result()
epochs = len(results['validation_0']['logloss'])
x_axis = range(0, epochs)

axes[0, 0].plot(x_axis, results['validation_0']['logloss'], label='Train Loss', color='teal')
axes[0, 0].plot(x_axis, results['validation_1']['logloss'], label='Validation Loss', color='coral')
axes[0, 0].set_title('XGBoost LogLoss Convergence Graph')
axes[0, 0].set_xlabel('Number of Iterations (Trees)')
axes[0, 0].set_ylabel('LogLoss Logarithmic Scale')
axes[0, 0].legend()

#confusion matrix heatmap
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0, 1], cbar=False)
axes[0, 1].set_title('Confusion Matrix Summary')
axes[0, 1].set_xlabel('Predicted Label Class')
axes[0, 1].set_ylabel('True Actual Class')

#xgboost feature important
importance = model.get_booster().get_score(importance_type='gain')
importance_df = pd.DataFrame({
    'Feature': list(importance.keys()),
    'Gain': list(importance.values())
}).sort_values(by='Gain', ascending=False)

sns.barplot(x='Gain', y='Feature', data=importance_df, palette='viridis', ax=axes[1,0])
axes[1,0].set_title('Feature Importance Framework (Gain)')
axes[1,0].set_xlabel('Relative Splits Improvement Gain Metric')

#receiver operating charecteristic (roc) profile
fpr, tpr, _ = roc_curve(y_test, y_prob)
axes[1,1].plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc_score(y_test, y_prob):.3f})')
axes[1,1].plot([0, 1], [0,1], color='navy', lw=2, linestyle='--')
axes[1,1].set_xlim([0.0, 1.0])
axes[1,1].set_ylim([0.0, 1.05])
axes[1,1].set_title('Receiver Operating Characteristic Profile')
axes[1,1].set_xlabel('False Positive Evaluation Rate')
axes[1,1].set_ylabel('True Positive Evaluation Rate')
axes[1,1].legend(loc="lower right")

plt.tight_layout()
plt.show()
```

step 7
