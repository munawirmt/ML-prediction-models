# **EDA & FEATURE ENGINEERING (BLACK FRIDAY DATASET)**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
## load train dataset from drive
from google.colab import drive
drive.mount("/content/drive")
file_path="/content/drive/MyDrive/black friday dataset/train-selected-columns.csv"
df_train=pd.read_csv(file_path)
df_train.head()
```

```python
##import test dataset from drive
file_path="/content/drive/MyDrive/black friday dataset/test-selected-columns.csv"
df_test=pd.read_csv(file_path)
df_test.head()
```

```python
## merge train and test datasets
df= pd.concat([df_train, df_test], ignore_index=True)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
## remove user_id
print("Current columns in df:", df.columns.tolist())
# Strip whitespace from column names to handle potential invisible characters
df.columns = df.columns.str.strip()
if 'User_ID' in df.columns:
    df.drop(['User_ID'], axis=1, inplace=True)
    print("User_ID column dropped successfully.")
else:
    print("User_ID column not found after stripping whitespace. Please check the exact column name.")
```

```python
df.head()
```

```python
## handle categorical feature gender
df['Gender']= df['Gender'].map({'F':0, 'M':1})
```

```python
## Combine Product_Category_1 and Product_Category_2 into a 'Purchase' column
# Assuming 'combine' means sum, as both are float types, but they are currently all NaN.
df['Purchase'] = df['Product_Category_1'] + df['Product_Category_2']

# Visualize 'Purchase' vs 'Age' barplot
import matplotlib.pyplot as plt
import seaborn as sns

if 'Age' in df.columns and not df['Age'].isnull().all() and \
   'Purchase' in df.columns and not df['Purchase'].isnull().all():
    plt.figure(figsize=(10, 6))
    sns.barplot(x='Age', y='Purchase', data=df, errorbar=None)
    plt.title('Average Purchase by Age Group')
    plt.xlabel('Age Group')
    plt.ylabel('Average Purchase')
    plt.show()
else:
    print("Cannot create barplot: 'Age' or 'Purchase' column is missing or entirely null. Please re-check data loading and column content.")
```

```python
## handle categorical feature city_category
df_city = pd.get_dummies(df['City_Category'], drop_first=True)
df_city.head()
```

```python
## add both
df = pd.concat([df, df_city], axis=1)
df.head()
```

```python
## delete city category
df.drop(['City_Category'], axis=1, inplace=True)
df.head()
```

```python
## missing values
df.isnull().sum()
```

```python
## focus on replacing missing values (category 1,2)
df['Product_Category_1'].unique() ## descrete feature
df['Product_Category_2'].unique() ## descrete feature

df['Product_Category_2'].value_counts()
```

```python
## replace missing values with mode (best way to descrete feature)
df['Product_Category_2'].mode()
df['Product_Category_2'] = df['Product_Category_2'].fillna(df['Product_Category_2'])
df['Product_Category_2'].isnull().sum()
```

```python
df['Product_Category_1'].mode()
df['Product_Category_1'] = df['Product_Category_2'].fillna(df['Product_Category_1'])
df['Product_Category_1'].isnull().sum()
```
