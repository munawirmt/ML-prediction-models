```python
drive.mount("/content/drive")
file_path = "/content/drive/MyDrive/Zomatodataset/Country-Code.xlsx"
df_country=pd.read_excel(file_path)
df_country.head()
```

```python
##combine tow datasets (zomato.csv & country-code.xlsx)
final_df=pd.merge(df,df_country,on='Country Code',how='left')
final_df.head()
```

```python
##check datatypes
final_df.dtypes
```

```python
final_df.columns
```

```python
final_df.Country.value_counts()
```

```python
country_names=final_df.Country.value_counts().index
country_val=final_df.Country.value_counts().values
```

```python
##pie chart
plt.pie(country_val,labels=country_names,autopct='%1.2f%%')
```

```python
##top 3 countries pie chart
plt.pie(country_val[:3],labels=country_names[:3],autopct='%1.2f%%')
```

**observation:**
    zomato maximum records and transactions from india , usa , uk


```python
final_df.columns
```

```python
##taking main 3 features and see the relationship between them
ratings=final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
ratings
```

**Observation:**

1.   when rating is between 4.5 to 4.9 -> excellent
2.   when rating is between 4 to 4.4 -> very good
3.   when rating is between 3.5 to 3.9 -> good
4.   when rating is between 3 to 3.4 -> average
5.   when rating is between 2.5 to 2.9 -> average
6.   when rating is between 2 to 2.4 -> poor

  






```python
ratings.head()
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize']=(12,6)
sns.barplot(x='Aggregate rating',y='Rating Count',data=ratings)
```

```python
sns.barplot(x='Aggregate rating',y='Rating Count',hue='Rating color',data=ratings,palette=['purple','red','orange','yellow','green','darkgreen'])
```

**Observation**:



1.   Not rated count is very high
2.   maximum rating is between 2.5 to 3.4
3.   if some rating are missing use average (2.5 to 3.4)


```python
##count plot
sns.countplot(x='Rating color',data=ratings,palette=['purple','red','orange','yellow','green','darkgreen'])
```
